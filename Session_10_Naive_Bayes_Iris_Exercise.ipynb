## 1. What is the Naive Bayes Classifier?

Naive Bayes is a machine learning classification algorithm that predicts the category of a data point using probability.

Bayes’ Theorem


$P(B|A) = \frac{P(A|B) \cdot P(B)}{P(A)}$

Explanation of Terms
- Posterior (P(B|A))  
  - The probability of the hypothesis B being true given the observed data A.  

- Likelihood (P(A|B))  
  - The probability of observing the data A assuming the hypothesis B is true.  
  - It measures how well the hypothesis explains the observed evidence.

- Prior (P(B))  
  - The initial probability of the hypothesis B before seeing any data.  
  - It represents our knowledge about B in advance.

- Evidence (P(A))  
  - The overall probability of observing the data A, regardless of the hypothesis.  
  - It acts as a normalizing constant to ensure the posterior probabilities sum to 1 across all possible hypotheses.

-----------------------------------------------------------------

Conditional Independence Assumption

The classifier assumes that **all features are conditionally independent given the class label**.  
Formally, if we have features $(x_1, x_2, \dots, x_n)$ and a class $(y)$, Naive Bayes assumes:

$
P(A_1, A_2, \dots, A_n | B) = P(A_1 | B) . P(A_2 | B) \dots P(A_n | B)
$

This means: if the class \(B\) is known, the presence or value of one feature does not affect the probability of another.

----------------------------------------------------------------

Why is this assumption "naive"?
- In reality, features are often **correlated**. For example, in text classification, the words "machine" and "learning" tend to appear together — they are not independent.  
- By ignoring these correlations, the model oversimplifies the true structure of the data. That is why the assumption is considered "naive."

----------------------------------------------------------------

Why does Naive Bayes still work well?
- Even though the independence assumption is rarely true, the classifier often makes **accurate predictions** because:
  - It captures the **dominant signal** in the data.
  - Errors from the independence assumption may cancel out across features.
  - It is computationally efficient and works well with high-dimensional data (like text).

## 3. Why Gaussian Naive Bayes?

Why Gaussian Naive Bayes for the Iris dataset?

- The **Iris dataset** has four features: sepal length, sepal width, petal length, and petal width.  
- Each of these is a **continuous measurement in centimeters** (not counts, not binary flags).  
- **Gaussian Naive Bayes (GaussianNB)** assumes that, within each class (species of iris), the distribution of each continuous feature follows a **normal (Gaussian) distribution**.  
  - For example, petal length for *Iris setosa* might cluster around a mean value with some variance, forming a bell-shaped curve.

---

Why not MultinomialNB?
- **MultinomialNB** is designed for **count data** (e.g., word frequencies in text classification).  
- It models features as discrete counts (like 0, 1, 2 occurrences of a word).  
- Continuous measurements like “4.7 cm” don’t fit this framework.

---

 Why not BernoulliNB?
- **BernoulliNB** is for **binary features** (yes/no, present/absent).  
- It expects inputs like “feature present = 1” or “feature absent = 0.”  
- Again, continuous measurements like sepal length don’t match this assumption.

---

GaussianNB is appropriate because
- the Iris features are **continuous real-valued measurements**, GaussianNB is the natural choice.  
- it models each feature’s distribution per class as a Gaussian curve, which aligns well with the nature of the data.  
- This allows the classifier to estimate likelihoods $P(A_i | B)$ using the Gaussian probability density function.

---


- **GaussianNB** = continuous features (like Iris measurements).  
- **MultinomialNB** = count-based features (like word frequencies).  
- **BernoulliNB** = binary features (like yes/no flags).

Naive Bayes assumes that features are **conditionally independent** given the class.  
Let’s see how correlated the four Iris features actually are.

### Strongest correlations

The pairs with the highest absolute correlations are usually:

- Petal length ↔ Petal width  (very strong positive correlation)
- Sepal length ↔ Petal length / Petal width

These features are clearly **not** independent.

### What effect does this have?

When features are strongly correlated, the likelihood $P(x \mid y)$ calculated by Naive Bayes is not exactly correct (it double-counts shared information). Consequently the **absolute probability values** can be over- or under-confident.

However, the **ranking** of the classes is often still correct. Because classification only needs the class with the largest posterior, the model can still make good predictions even when the independence assumption is violated. This is one of the reasons Naive Bayes remains a popular and effective baseline classifier.

## 7. Working Through One Prediction by Hand

### Gaussian probability density formula

$$
P(x \mid y) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)
$$

where $\mu$ is the class mean and $\sigma^2$ is the class variance that the model learned.

We will:
1. Pick one test sample
2. Take its petal-length value
3. Compute the likelihood under each of the three classes
4. Multiply by the class prior
5. Normalize so the three numbers sum to 1 (these become the posterior probabilities)
6. Compare the result with `model.predict_proba()`

```python


# Gaussian PDF function
def gaussian_pdf(x, mean, var):
    """Return the probability density of x under a Normal(mean, var) distribution."""
    return (1 / np.sqrt(2 * np.pi * var)) * np.exp(-((x - mean) ** 2) / (2 * var))
```

## 8. Comparing Naive Bayes with Logistic Regression

Let’s train a **Logistic Regression** model on the same train/test split and compare the two classifiers.

Important difference: Logistic Regression benefits from **feature scaling**, while GaussianNB does not need it.

```python
# import library
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
```

```python
# Scale the features (important for Logistic Regression)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)
```

```python
# Train Logistic Regression
lr = LogisticRegression(max_iter=200, random_state=42)
lr.fit(X_train_scaled, y_train)
```

```python
# Predictions
y_pred_lr = lr.predict(X_test_scaled)
```

```python
# Collect metrics for both models
def get_metrics(y_true, y_pred):
    return {
        "Accuracy":  accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, average="macro"),
        "Recall":    recall_score(y_true, y_pred, average="macro"),
        "F1-score":  f1_score(y_true, y_pred, average="macro")
    }

metrics_nb = get_metrics(y_test, y_pred)
metrics_lr = get_metrics(y_test, y_pred_lr)
```

```python


# summary table
summary = pd.DataFrame({
    "GaussianNB": metrics_nb,
    "Logistic Regression": metrics_lr
}).round(3)

print("Comparison of the two models (macro-averaged metrics):\n")
print(summary)
```

### What does this comparison tell us?

**Advantages of Gaussian Naive Bayes**
- No feature scaling required
- No hyperparameter tuning needed for a basic model
- Extremely fast to train (just computes means, variances and priors)
- Works well even with small datasets

**When you might still prefer Logistic Regression**
- When features are strongly correlated (LR can learn the relationships; NB assumes independence)
- When you need well-calibrated probability estimates
- When you want to interpret the effect of each feature through coefficients
- When the decision boundary is more complex than what independent Gaussians can capture

In short: **GaussianNB is a fantastic baseline** — simple, fast and often surprisingly accurate. Logistic Regression (and other more flexible models) become preferable when you have enough data and need higher accuracy or better probability calibration.

## Summary

- Naive Bayes is a simple, fast probabilistic classifier based on Bayes’ theorem.
- The word “naive” refers to the strong assumption that features are independent given the class.
- For continuous features we use **GaussianNB**, which models each feature as a normal distribution inside each class.
- On the Iris dataset the model achieves high accuracy; the few mistakes usually occur between versicolor and virginica – the two species that overlap most in feature space.
- Even though the features are correlated (violating the independence assumption), the classifier still works well because we only need the correct ranking of class probabilities.
- You can compute the posteriors by hand using the Gaussian PDF and the learned means/variances — this helps you understand exactly what the model is doing.
- Compared with Logistic Regression, GaussianNB needs no scaling and almost no tuning, making it extremely convenient. Logistic Regression can still be preferred when feature interactions matter or when better-calibrated probabilities are required.
